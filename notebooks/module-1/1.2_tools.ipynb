## Tool definition

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain.tools import tool

@tool
def square_root(x: float) -> float:
    """Calculate the square root of a number"""
    return x ** 0.5
```

```python
@tool("square_root")
def tool1(x: float) -> float:
    """Calculate the square root of a number"""
    return x ** 0.5
```

```python
@tool("square_root", description="Calculate the square root of a number")
def tool1(x: float) -> float:
    return x ** 0.5
```

```python
tool1.invoke({"x": 467})
```

## Adding to agents

```python
from langchain.agents import create_agent

agent = create_agent(
    model = "google_genai:gemini-2.5-flash",
    tools =  [tool1],
    system_prompt = "You are a math teacher. What you need to do is show me the answer of some math questions, and you need to tell me the process of calculation."
)
```

```python
from langchain.messages import HumanMessage
from pprint import pprint

question = HumanMessage(content = "what is the result of 1 * 2 + sqrt(9) - 3?")
response = agent.invoke({"messages": [question]})

pprint(response['messages'])
```
